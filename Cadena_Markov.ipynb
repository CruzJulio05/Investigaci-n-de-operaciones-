```python
import random as r
import sympy as s
```

La función random() devuelve un número aleatorio $x_i $~$ U{(0,1)}$

```python
r.random()
```

Para una moneda justa $P(S)=0.5$ y $P(A)=0.5$

Tomamos 1 para Sol; 0 para Águila

```python
def volado(n):
   lista=[]
   for i in range(n):
      v=r.random()
      if v <= 0.5:
         x=1
      else:
         x=0
      lista.append(x)
   return lista
```

```python
volado(20)
```
